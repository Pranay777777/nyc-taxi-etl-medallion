# 01 — Landing Layer
**Purpose:** verify raw Parquet has landed in the source volume; do NOT transform.  
**Source:** NYC TLC Yellow Taxi, January 2024     
**Volume path:** `/Volumes/workspace/taxi/raw/`

```python
from pyspark.sql import functions as F

SOURCE_PATH = "/Volumes/workspace/taxi/raw/yellow_tripdata_2024-01 (2).parquet"

# Confirm the file is readable and check what we're getting
df_raw = spark.read.parquet(SOURCE_PATH)

print(f"Source file: {SOURCE_PATH}")
print(f"Row count:   {df_raw.count():,}")
print(f"Column count: {len(df_raw.columns)}")
print("\n--- Schema ---")
df_raw.printSchema()
print("\n--- First 5 rows ---")
df_raw.show(5, truncate=False)
```
